## 6) Simple linear regression
Goal: predict **price (thousands of PLN)** from **floor area (m²)**.
Steps:
1. Split features X (area) and target y (price).
2. Fit `LinearRegression`.
3. Inspect parameters (coefficient and intercept).
4. Plot the fitted line.


````python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

# X: floor area (m²), y: price (thousands of PLN)
X = df[["area_m2"]].values
y = df["price_thousands_pln"].values

model = LinearRegression()
model.fit(X, y)

print("Coefficient (a):", model.coef_[0])
print("Intercept (b):", model.intercept_)

# Predictions for a few sample areas
sample = np.array([[40],[60],[80],[100]])
pred = model.predict(sample)
for m2, p in zip(sample.flatten(), pred):
    print(f"Predicted price for {m2} m²: {p:.1f} thousand PLN")

# Plot: points + regression line
plt.figure()
plt.scatter(X, y, label="Data")
x_line = np.linspace(X.min(), X.max(), 100).reshape(-1,1)
y_line = model.predict(x_line)
plt.plot(x_line, y_line, label="Linear regression")
plt.xlabel("Floor area (m²)")
plt.ylabel("Price (thousands of PLN)")
plt.title("Simple linear regression: price = a*area + b")
plt.legend()
plt.show()

# Exercise 1: Calculate price per m²
df["price_per_m2"] = (df["price_thousands_pln"] * 1000) / df["area_m2"]
print(df[["area_m2", "price_thousands_pln", "price_per_m2"]].head())

print("\nPrice per m² summary:")
print(df["price_per_m2"].describe())

# Exercise 2: Add squared area
X_poly = np.hstack([X, X**2])

model_poly = LinearRegression()
model_poly.fit(X_poly, y)
y_pred_poly = model_poly.predict(X_poly)

# Compute MSE
mse_linear = np.mean((y - model.predict(X)) ** 2)
mse_poly = np.mean((y - y_pred_poly) ** 2)

print(f"Simple Linear Regression MSE: {mse_linear:.2f}")
print(f"Polynomial Regression MSE: {mse_poly:.2f}")

# Exercise 3: Save plot
plt.figure(figsize=(8, 5))
plt.scatter(X, y, label="Data", color="blue")
x_line = np.linspace(X.min(), X.max(), 100).reshape(-1, 1)
y_line = model.predict(x_line)
plt.plot(x_line, y_line, label="Linear regression", color="red")

plt.xlabel("Floor area (m²)")
plt.ylabel("Price (thousands of PLN)")
plt.title("Simple linear regression: price = a*area + b")
plt.legend()

plt.savefig("regression.png", dpi=300, bbox_inches="tight")
print("Plot saved as regression.png")
plt.show()

# Extension: Train/test split and error metrics
np.random.seed(42)
indices = np.random.permutation(len(X))
split_idx = int(0.8 * len(X))

# Split indices into train and test
train_idx, test_idx = indices[:split_idx], indices[split_idx:]
X_train, X_test = X[train_idx], X[test_idx]
y_train, y_test = y[train_idx], y[test_idx]

# Train model
model_split = LinearRegression()
model_split.fit(X_train, y_train)
y_pred_test = model_split.predict(X_test)

# Compute MAE and RMSE manually
mae = np.mean(np.abs(y_test - y_pred_test))
rmse = np.sqrt(np.mean((y_test - y_pred_test) ** 2))

print(f"Test MAE: {mae:.2f} thousand PLN")
print(f"Test RMSE: {rmse:.2f} thousand PLN")
````
